```python
def calculate_percentiles(data):
    sorted_data = sorted(data)
    sorted_data = list(map(int,sorted_data))
    total_items = len(sorted_data)
    percentiles = [i / 10 for i in range(10, 101, 10)]  # 10%, 20%, ..., 100%
    result = {}

    for percentile in percentiles:
        index = int(percentile * total_items * 0.1)
        # インデックスがリストの範囲内かどうかをチェック
        if 0 <= index - 1 < total_items:
            result[f'{int(percentile * 100)}%'] = sorted_data[index - 1]
        else:
            result[f'{int(percentile * 100)}%'] = None  # 範囲外の場合 None にする

    return result

numbers = df_kamoku["CODE_NUM"].astype(str).tolist()
code_num_list = list(map(str,df_kamoku["CODE_NUM"]))
# nendo_list = df_kamoku["CODE_NUM"].astype(str).tolist.str[-4:-2]
top_two_digits_list = []
bottom_two_digits_list = []

# 各要素に対して上二桁と下二桁を取り出してリストに追加
for num in numbers:
    top_two_digits_list.append(str(num)[-4:-2])
    bottom_two_digits_list.append(str(num)[-2:])

for i,j in zip(top_two_digits_list,bottom_two_digits_list):

  df['KAMOKU_ID'] = df['ENSHU_SET_ID'].str[4:6]
  df['SHIKEN_NENDO'] = df['ENSHU_SET_ID'].str[-2:]
  df['DAIMON_NUM'] = df['DAIMON_ID'].str[-2:]
  df_try = df[(df["KOZA_CD"] == "1408")&(df["KAMOKU_ID"]=="10") & (df["SHIKEN_NENDO"]==i)&(df["DAIMON_NUM"]==j)]
  tokuten = df_try["DAIMON_TOKUTEN"].tolist()
  # print(tokuten)
  # print(i,j)

  # NaN を除外して、数値だけをリストに残す
  tokuten_cleaned = [x for x in tokuten if pd.notna(x)]  # NaN を除外

  # ここにDAIMON_TOKUTENの数値リストを入力
  data = list(map(lambda x: int(float(x)), tokuten_cleaned))  # floatに変換してからintに変換

  result = calculate_percentiles(data)
  list4 = list(result.values())
  print(list4)
  # for percentile, value in result.items():
  #     print(f'{percentile}の値: {value}')
```
